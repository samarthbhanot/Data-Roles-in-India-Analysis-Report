```python
df['job_title_clean'] = df['job_title'].astype(str).str.strip()
```

```python
df['job_title_clean'] = df['job_title_clean'].str.strip().replace(r'\s+', ' ', regex=True) \
                                         .str.replace('...', '', regex=False) \
                                         .str.strip() \
                                         .str.title()
df['job_title_clean'].head()
                                     
```

```python
df['job_title_clean'] = df['job_title_clean'].str.split('-').str[0]
df['job_title_clean'] = df['job_title_clean'].str.split('/').str[0]
df['job_title_clean'] = df['job_title_clean'].str.split(',').str[0]
```

```python
df['job_title_clean'].head()
```

```python
def std(t):
    t = t.lower()
    if 'intern' in t:
        return 'Intern'
    if 'fresher' in t:
        return 'Fresher'
    if 'senior' in t or 'lead' in t or 'avp' in t:
        return 'Senior / Lead'
    return 'Data Analyst'

df['job_level'] = df['job_title_clean'].apply(std)
df['job_level'].value_counts()
```

```python
def get_domain(t):
    t = t.lower()
    if 'agricultural' in t:
        return 'Agriculture'
    if 'health' in t or 'clinical' in t:
        return 'Healthcare'
    if 'finance' in t or 'fintech' in t or 'bank' in t:
        return 'Finance'
    if 'marketing' in t or 'sales' in t:
        return 'Marketing'
    if 'hr' in t or 'human resource' in t:
        return 'HR'
    if 'ecommerce' in t or 'retail' in t:
        return 'ECommerce / Retail'
    return 'General / IT'

df['job_domain'] = df['job_title'].apply(get_domain)
df['job_domain'].value_counts()
```

```python
full_text = ' '.join(df['job_description'].astype(str)).lower()
from collections import Counter
words = full_text.split()
common = Counter(words).most_common(50)
common
```

```python
my_words =['python', 'sql', 'excel', 'power bi', 'tableau', 'reporting', 'dashboard', 'analytics', 'insights', 'working',
           'team', 'analysis', 'skills', 'support', 'strong', 'reputed', 'experience', 'analysis', 'role', 'work']

def pick_words(text):
    text = str(text).lower()
    found = [w.title() for w in my_words if w in text]
    return ", ".join(found)
df['job_description_keywords'] = df['job_description'].apply(pick_words)
df[['job_title', 'job_description_keywords']].head()
```
